```python
df = df.dropna(subset=["Transaction_ID"])
```

```python
bad_dates = ["2023-13-01", "2025-02-30"]
bad_dates_count = df["Transaction_Date"].isin(bad_dates).sum()
print(f"Number of bad dates: {bad_dates_count}")
```

```python
df["Transaction_Date"]=df["Transaction_Date"].replace("2023-13-01","2024-01-01")
df["Transaction_Date"]=df["Transaction_Date"].replace("2025-02-30","2025-03-01")
```

```python
bad_dates_count = df["Transaction_Date"].isin(bad_dates).sum()
print(f"Number of bad dates: {bad_dates_count}")
df.shape
```

```python
df["Transaction_Date"]=pd.to_datetime(df["Transaction_Date"])
print(df["Transaction_Date"].dtype)
```

```python
((df["Transaction_Date"].isna().sum())*100)/(len(df))
```

```python
df=df.dropna(subset=["Transaction_Date"])
df.shape
```

```python
((df["Customer_ID"].isna().sum())*100)/(len(df))
```

```python
df=df.dropna(subset=["Customer_ID"])
df.shape
```

```python
df["Product_Name"]=df["Product_Name"].str.strip()
df["Product_Name"].value_counts()
```

```python
df["Product_Name"]=df["Product_Name"].replace(["Tabl","Ta","Table","T","Tab","Tablet"],"Tablet")
df["Product_Name"]=df["Product_Name"].replace(["Lapt","La","Lapto","Lap","L","Laptop"],"Laptop")
df["Product_Name"]=df["Product_Name"].replace(["Coffee","Coffee Machi","Coffee Ma","Coffee Machin","Coffee Mach","Coffee M","C","Cof","Coffe","Coffee Mac","Coff","Co","Coffee Machine"],"Coffee Machine") 
df["Product_Name"]=df["Product_Name"].replace(["Smar","Smartpho","Sma","Sm","Smart","Smartp","Smartphon","S","Smartph","Smartphone"],"Smartphone")
df["Product_Name"]=df["Product_Name"].replace(["Hea","Headph","Headphone","Headphon","Head","Headpho","He","Headp","H"],"Headphones")
```

```python
df["Product_Name"].value_counts()
```

```python
((df["Quantity"].isna().sum())*100)/(len(df))
```

```python
((df["Quantity"]<0).sum()*100)/(len(df))
```

```python
df['Quantity_Invalid'] = df['Quantity'] < 0
df.loc[df['Quantity'] < 0, 'Quantity'] = np.nan
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].median())
```

```python
df["Price"] = df["Price"].apply(lambda x: x.replace("$", "").strip() if isinstance(x, str) else x)
df["Price"] = pd.to_numeric(df["Price"], errors='coerce')
(df["Price"]).dtype
```

```python
((df["Price"].isna().sum())*100)/(len(df))
```

```python
((df["Price"]<0).sum()*100)/(len(df))
```

```python
df['Price_Invalid'] = df['Price'] < 0
df.loc[df['Price'] < 0, 'Price'] = np.nan
df['Price'] = df.groupby('Product_Name')['Price'].transform(
    lambda x: x.fillna(x.median())
)
df['Price'] = df['Price'].fillna(df['Price'].median())
```

```python
((df["Payment_Method"].isna().sum())*100)/(len(df))
```

```python
df["Payment_Method"]=df["Payment_Method"].str.strip()
df["Payment_Method"].value_counts()
```

```python
df["Payment_Method"]=df["Payment_Method"].replace(["Credit Card","creditcard","credit card"],"Credit Card")
df["Payment_Method"]=df["Payment_Method"].replace(["pay pal","PayPal","PayPal"],"Pay Pal")
```

```python
((df["Transaction_Status"].isna().sum())*100)/(len(df))
```

```python
df["Transaction_Status"]=df["Transaction_Status"].str.strip()
df['Status_Missing'] = df['Transaction_Status'].isna()
df["Transaction_Status"]=df["Transaction_Status"].replace(np.nan,"Unknown")
df["Transaction_Status"].value_counts()
```

```python
df["Transaction_Status"]=df["Transaction_Status"].replace(["complete","completed","Completed"],"Completed")
```
